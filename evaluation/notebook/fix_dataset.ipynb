```python
import pandas as pd
```

```python
df = pd.read_json("../datasets/papers-data.json")
```

```python
multi_category_df = df[
    df.groupby("title")["title"].transform("count") > 1
]

multi_category_df.sort_values(by="title")
```

```python
def merge_categories(s: pd.Series) -> str:
    cats = set(s.dropna().astype(str).str.strip())
    return ", ".join(cats)
```

```python
out = (
    df.groupby("link", as_index=False)
      .agg({
          "title": "first",
          "authors": "first",
          "year": "first",
          "abstract": "first",
          "category": merge_categories,
      })
)

out.shape
```

```python
import json

records = out.to_dict(orient="records")

with open("..datasets/fixed_dataset.json", "w", encoding="utf-8") as f:
    json.dump(records, f, ensure_ascii=False, indent=4)
```
